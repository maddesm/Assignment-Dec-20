```python
# Write a python program that computes the factorial of a number taken from the user :
# for example 4 factorial : 4 * 3 * 2 * 1

n = int(input())
result = 1
# why should you put n+1 instead of n ?
for i in range(1,n+1):
    result = i * result
print(result)
```
